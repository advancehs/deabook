```
from deabook import PANEL
# 导入Malmquist指数分析所需包
from deabook import DEA
from deabook.constant import RTS_CRS, \
                        OPT_LOCAL, CONTEMPORARY  # 导入DEA分析常量

import pandas as pd
# 导入所有数据（包括环境变量）
from pandas.api.types import CategoricalDtype  # 用于创建分类数据类型

# 定义Malmquist指数分解成分的顺序
cat_size_order = CategoricalDtype(
    ['MEFFCH', 'MTECHCH', 'MQ'],  # Malmquist指数分解项顺序：效率变化、技术变化、总指数
    ordered=True  # 保持指定顺序
)

# 读取 OECD 国家数据，筛选 2015 年后观测
dmuname = '省份'  # 决策单元名称列
data = (
    pd.read_excel(data_path("china data.xlsx"))
    .query('year>2015')
    .reset_index(drop=True)
)

# 提取年份和决策单元标识
yearid = data[['year', dmuname]]  
# 查看数据中包含的决策单元
yearid[dmuname].unique()  

# 生成投入产出变量的描述性统计表（Y=产出，K=资本，L=劳动，E=能源，CO2=碳排放）
data[['Y','K','L','E','CO2']].describe().T.to_markdown("table4.11.md")
```

```
from deabook import PANEL
# 定义并计算基于CRS的Malmquist生产率指数（FGNZ 1994 式）
def mq_crs(data, dmuname):
    """
    构建Malmquist DEA模型（规模报酬不变/CRS假设）
    - data: 包含投入产出数据的DataFrame
    - id=dmuname: 决策单元标识列名
    - sent="K L E=Y": 资本、劳动、能源为投入，Y为产出
    - orient="oo": 产出导向
    - rts=RTS_CRS: 规模报酬不变
    - tech=CONTEMPORARY: 同期技术前沿（默认）
    - email/solver 传给 .optimize(...)
    """
    model = DEA.mq(data, id=dmuname, year='year',
                   sent="K L E=Y",
                   orient="oo",   # 产出导向
                   rts=RTS_CRS,  # 规模报酬不变（CRS）
                   tech=CONTEMPORARY,  # 同期技术前沿
                   )
    return model.optimize(email=OPT_LOCAL, solver="mosek")

# 执行模型计算
res_crs = mq_crs(data, dmuname)

# 提取关键结果列（年份、DMU名称、M指数及其分解项）
res_crs2 = res_crs[['year', dmuname, 'MQ', 'MEFFCH', 'MTECHCH']]

# 创建透视表（按DMU和年份展示结果）
res_crs3 = res_crs2.query('year>2016').pivot_table(
    values=['MQ', 'MEFFCH', 'MTECHCH'],
    index=dmuname,
    columns='year',
    aggfunc='mean'
)[['MEFFCH', 'MTECHCH', 'MQ']]  # 调整列顺序

# 保存结果为Markdown表格
res_crs3.to_markdown("table4.12_mq_CRS.md")
```

```
from deabook import PANEL
# 定义方向距离函数(DDF)的Malmquist模型（产出导向，g=(0,y)）
def mqddf_crs(data, dmuname):
    """
    构建基于方向距离函数的Malmquist模型（CRS假设）
    - gx=[0,0,0], gy=[1]：产出方向 g=(0,y)，底层 teo=1/(1+rho)（见 tbl-3-6）
    - tech=CONTEMPORARY：同期技术前沿（默认）
    """
    model = DEA.mqddf(data, id=dmuname, year='year',
                      sent="K L E=Y",
                      gx=[0, 0, 0], gy=[1],
                      rts=RTS_CRS,
                      tech=CONTEMPORARY,
                      )
    return model.optimize(email=OPT_LOCAL, solver="mosek")

# 执行模型并处理结果
res_ddf = mqddf_crs(data, dmuname)
res_ddf2 = res_ddf[['year', dmuname, 'MQ', 'MEFFCH', 'MTECHCH']]

# 生成透视表并保存结果
res_ddf3 = res_ddf2.query('year>2016').pivot_table(
    values=['MQ', 'MEFFCH', 'MTECHCH'],
    index=dmuname,
    columns='year'
)[['MEFFCH', 'MTECHCH', 'MQ']]  # 调整列顺序
res_ddf3.to_markdown("table4.13_mqddf_CRS.md")
```
